```python
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
print(categorical_cols)
```

```python
print("Target variable distribution:")
print(df['responded'].value_counts())
print("\nPercentage distribution:")
print((df['responded'].value_counts(normalize=True) * 100).round(2))
```

```python
missing_train = df.isnull().sum()
print("Missing values in training data:")
print(missing_train)
```

```python
missing_df = pd.DataFrame({
    'Missing Count': missing_train,
    'Missing Percentage': (missing_train / len(df) * 100).round(2)
})  
missing_df[missing_df['Missing Count'] > 0].sort_values(by='Missing Count', ascending=False)
```
